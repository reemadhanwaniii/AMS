```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate,ChatPromptTemplate
```

```python


llm = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen2.5-1.5B-Instruct",
    task="text-generation"
)
```

```python
prompt = "What is merge sort algorithm ? please write code for it as well"

response = llm.invoke(prompt)
print(response)
```

```python
from langchain_huggingface import HuggingFacePipeline

llm = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen2.5-1.5B-Instruct",
    task="text-generation",
    pipeline_kwargs={
        'max_new_tokens' : 2048
    }
)


prompt = "what is insertion sort? please write code for that as well"
response = llm.invoke(prompt)
print(response)
```

```python
llmdemo = HuggingFacePipeline.from_model_id(
    model_id="crumb/nano-mistral",
    task="text-generation",
    pipeline_kwargs={
        'max_new_tokens': 200
    }
)

greedy = "what is greedy algorithms?Give me some code examples"
response_crumb = llmdemo.invoke(greedy)
print(response_crumb)
```

promptTemplates :
-- prompt templates are reusable instructions which can be passed to langchain using different params

```python
cs_tutor_template = """
You are a expert computer science engineer. Explain the following algorithm : {algorithm_name}

FOLLOW THE EXACT STRUCTURE:

### 1. The Concept(what it is?)
Explain what problem this algorithm solves in 2 sentences

### 2. The efficiency (Complexity analysis)
Define the time complexity and space complexity of the {algorithm_name}

### 3. Sample trace run
Show a simple step by step dry run using the sample input : {sample_input}

Answer::

"""
```

```python
tutor_prompt = PromptTemplate(
    template=cs_tutor_template,
    input_variables=["algorithm_name","sample_input"]
)
```

```python
final_prompt = tutor_prompt.format(algorithm_name="Bubble Sort",sample_input="[5,6,3,2,1,7,8]")
```

```python
print(final_prompt)
```

```python
chain = tutor_prompt | llm # prompt ----> feeded inside llm ---> llm
```

```python
result = chain.invoke({
    "algorithm_name":"Binary Search",
    "sample_input": "Target 7 in list [6,7,8,3,2,9]"
})

print(result)
```

```python
template = ChatPromptTemplate.from_messages(
    [
        ("system","You are a calculator which responds with reasoning about calculation"),
        ("human","Answer what is 10 multiplied by 20"),
        ("ai","10 x 20 = 200, as it's equivalent to adding 10 to itself 20 times"),
        ("human","{question}")
    ]
)
```

```python
chain = template | llm
response = chain.invoke({"question":"what is square root of 87"})
print(response)
```
